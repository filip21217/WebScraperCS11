# Packages import

```python
import  os
import datetime
import pandas as pd
from icalendar import Calendar, Event
```

# Calendar export

```python
group_name= input("enter group name: ")
```

```python
classes = pd.read_csv(f"schedules/{group_name}.csv",  encoding="UTF-8")
```

```python
classes
```

```python
if not os.path.exists("calendars"):
    os.mkdir("calendars")
```

```python
calendar = Calendar()
calendar['dtstart'] = "20260223T000000"
calendar['dtend'] = "20260930T235999"
calendar['summary'] = f"Schedule of the {group_name} group. "
```

```python
for index, row in classes.iterrows():
    event = Event()
    event['dtstart'] = row["Termin"].replace("-", "") + row["Start time"].replace(":", "")+"00"
    event['dtend'] = row["Termin"].replace("-", "") + row["Start time"].replace(":", "")+"00"
    event["location"] = row["Sala"]
    event["summary"] = row["Przedmiot"]
    calendar.add_component(event)
```

```python
print(calendar.to_ical())
```

```python
cf = open(f'calendars/{group_name}.ics', "w", encoding="UTF-8")
```

```python
cf.close()
```
